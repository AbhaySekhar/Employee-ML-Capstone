```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline

print("Preprocessing libraries imported successfully!")
```

```python
df_selected = pd.read_csv("../data/employee_hr_selected.csv")

print("Shape:", df_selected.shape)
print(df_selected.head())
```

```python
numerical_features = [
    "Age",
    "DistanceFromHome",
    "Education",
    "JobLevel",
    "NumCompaniesWorked",
    "TotalWorkingYears",
    "YearsAtCompany",
    "YearsInCurrentRole"
]

categorical_features = [
    "BusinessTravel",
    "Department",
    "JobRole",
    "OverTime"
]

print("Numerical features:", len(numerical_features))
print("Categorical features:", len(categorical_features))
print(
    "Total input features:",
    len(numerical_features) + len(categorical_features)
)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_features),
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        )
    ]
)

print("Preprocessor created successfully!")
```

```python
# Input features
X_reg = df_selected[
    numerical_features + categorical_features
].copy()

# Target variable
y_reg = df_selected["MonthlyIncome"].copy()

print("Features shape:", X_reg.shape)
print("Target shape:", y_reg.shape)

print("\nTarget preview:")
print(y_reg.head())
```

```python
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg,
    y_reg,
    test_size=0.2,
    random_state=42
)

print("Training features:", X_train_reg.shape)
print("Testing features:", X_test_reg.shape)

print("Training targets:", y_train_reg.shape)
print("Testing targets:", y_test_reg.shape)
```

```python
from sklearn.linear_model import LinearRegression

regression_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", LinearRegression())
    ]
)

regression_pipeline.fit(X_train_reg, y_train_reg)

print("Regression pipeline trained successfully!")
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

y_pred_reg = regression_pipeline.predict(X_test_reg)

mae = mean_absolute_error(y_test_reg, y_pred_reg)
mse = mean_squared_error(y_test_reg, y_pred_reg)
rmse = np.sqrt(mse)
r2 = r2_score(y_test_reg, y_pred_reg)

print("Regression Model Evaluation")
print("---------------------------")
print(f"MAE:  {mae:.2f}")
print(f"MSE:  {mse:.2f}")
print(f"RMSE: {rmse:.2f}")
print(f"R²:   {r2:.4f}")
```

```python
plt.figure(figsize=(7, 5))

plt.scatter(y_test_reg, y_pred_reg, alpha=0.7)

# Perfect prediction reference line
min_value = min(y_test_reg.min(), y_pred_reg.min())
max_value = max(y_test_reg.max(), y_pred_reg.max())

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    linestyle="--"
)

plt.xlabel("Actual Monthly Income")
plt.ylabel("Predicted Monthly Income")
plt.title("Actual vs Predicted Monthly Income")
plt.tight_layout()
plt.show()
```

```python
from sklearn.dummy import DummyRegressor

baseline = DummyRegressor(strategy="mean")

baseline.fit(X_train_reg, y_train_reg)

baseline_pred = baseline.predict(X_test_reg)

baseline_mae = mean_absolute_error(
    y_test_reg, baseline_pred
)

baseline_r2 = r2_score(
    y_test_reg, baseline_pred
)

print("Baseline MAE:", round(baseline_mae, 2))
print("Baseline R²:", round(baseline_r2, 4))

print("\nLinear Regression MAE:", round(mae, 2))
print("Linear Regression R²:", round(r2, 4))
```

```python
# Input features: all selected columns except Attrition
X_clf = df_selected.drop(columns=["Attrition"]).copy()

# Target variable
y_clf = df_selected["Attrition"].copy()

print("Features shape:", X_clf.shape)
print("Target shape:", y_clf.shape)

print("\nTarget distribution:")
print(y_clf.value_counts())
```

```python
X_train_clf, X_test_clf, y_train_clf, y_test_clf = train_test_split(
    X_clf,
    y_clf,
    test_size=0.2,
    random_state=42,
    stratify=y_clf
)

print("Training features:", X_train_clf.shape)
print("Testing features:", X_test_clf.shape)

print("\nTraining target percentages:")
print((y_train_clf.value_counts(normalize=True) * 100).round(2))

print("\nTesting target percentages:")
print((y_test_clf.value_counts(normalize=True) * 100).round(2))
```

```python
numerical_features_clf = [
    "Age",
    "DistanceFromHome",
    "Education",
    "JobLevel",
    "MonthlyIncome",
    "NumCompaniesWorked",
    "TotalWorkingYears",
    "YearsAtCompany",
    "YearsInCurrentRole"
]

categorical_features_clf = [
    "BusinessTravel",
    "Department",
    "JobRole",
    "OverTime"
]

preprocessor_clf = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_features_clf),
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features_clf
        )
    ]
)

print("Classification preprocessor created successfully!")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

classification_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor_clf),
        (
            "model",
            LogisticRegression(
                max_iter=2000,
                random_state=42
            )
        )
    ]
)

classification_pipeline.fit(X_train_clf, y_train_clf)

print("Classification model trained successfully!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    roc_auc_score
)

y_pred_clf = classification_pipeline.predict(X_test_clf)

y_proba_clf = classification_pipeline.predict_proba(X_test_clf)

# Probability of the positive class: Attrition = Yes
yes_index = list(
    classification_pipeline.named_steps["model"].classes_
).index("Yes")

y_proba_yes = y_proba_clf[:, yes_index]

print("Accuracy:", round(
    accuracy_score(y_test_clf, y_pred_clf), 4
))

print("\nClassification Report:")
print(classification_report(y_test_clf, y_pred_clf))

print("Confusion Matrix:")
print(confusion_matrix(
    y_test_clf,
    y_pred_clf,
    labels=["No", "Yes"]
))

print("ROC-AUC:", round(
    roc_auc_score(
        (y_test_clf == "Yes").astype(int),
        y_proba_yes
    ), 4
))
```

```python
# Create a class-weighted classification pipeline
balanced_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor_clf),
        (
            "model",
            LogisticRegression(
                max_iter=2000,
                class_weight="balanced",
                random_state=42
            )
        )
    ]
)

# Train the model
balanced_pipeline.fit(X_train_clf, y_train_clf)

# Make predictions
y_pred_balanced = balanced_pipeline.predict(X_test_clf)
y_proba_balanced = balanced_pipeline.predict_proba(X_test_clf)

# Get probability for Attrition = Yes
yes_index = list(
    balanced_pipeline.named_steps["model"].classes_
).index("Yes")

y_proba_yes_balanced = y_proba_balanced[:, yes_index]

# Evaluate
print("Balanced Model Accuracy:",
      round(accuracy_score(y_test_clf, y_pred_balanced), 4))

print("\nClassification Report:")
print(classification_report(y_test_clf, y_pred_balanced))

print("Confusion Matrix:")
print(confusion_matrix(
    y_test_clf,
    y_pred_balanced,
    labels=["No", "Yes"]
))

print("ROC-AUC:",
      round(roc_auc_score(
          (y_test_clf == "Yes").astype(int),
          y_proba_yes_balanced
      ), 4))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import precision_score, recall_score, f1_score

# Split the training data into training and validation sets
X_train_sub, X_val, y_train_sub, y_val = train_test_split(
    X_train_clf,
    y_train_clf,
    test_size=0.2,
    random_state=42,
    stratify=y_train_clf
)

# Train the balanced model on the smaller training set
balanced_pipeline.fit(X_train_sub, y_train_sub)

# Get probability of Attrition = Yes on validation data
yes_index = list(
    balanced_pipeline.named_steps["model"].classes_
).index("Yes")

val_probabilities = balanced_pipeline.predict_proba(X_val)[:, yes_index]

# Compare different thresholds
for threshold in [0.3, 0.4, 0.5, 0.6]:
    val_predictions = np.where(
        val_probabilities >= threshold, "Yes", "No"
    )

    precision = precision_score(
        y_val, val_predictions, pos_label="Yes", zero_division=0
    )
    recall = recall_score(
        y_val, val_predictions, pos_label="Yes", zero_division=0
    )
    f1 = f1_score(
        y_val, val_predictions, pos_label="Yes", zero_division=0
    )

    print(f"\nThreshold: {threshold}")
    print(f"Precision: {precision:.3f}")
    print(f"Recall:    {recall:.3f}")
    print(f"F1-score:  {f1:.3f}")
```

```python
# Retrain using the complete training set
balanced_pipeline.fit(X_train_clf, y_train_clf)

# Predict probabilities on the test set
yes_index = list(
    balanced_pipeline.named_steps["model"].classes_
).index("Yes")

test_probabilities = balanced_pipeline.predict_proba(
    X_test_clf
)[:, yes_index]

# Apply the selected threshold
y_pred_final = np.where(
    test_probabilities >= 0.6, "Yes", "No"
)

# Evaluate final predictions
print("Final Classification Report:")
print(classification_report(
    y_test_clf,
    y_pred_final,
    labels=["No", "Yes"],
    zero_division=0
))

print("Confusion Matrix:")
print(confusion_matrix(
    y_test_clf,
    y_pred_final,
    labels=["No", "Yes"]
))

print("Accuracy:",
      round(accuracy_score(y_test_clf, y_pred_final), 4))

print("ROC-AUC:",
      round(roc_auc_score(
          (y_test_clf == "Yes").astype(int),
          test_probabilities
      ), 4))
```

```python
#Unsupervised clustering
from sklearn.preprocessing import StandardScaler

# Select numerical features for clustering
clustering_features = [
    "Age",
    "DistanceFromHome",
    "JobLevel",
    "MonthlyIncome",
    "TotalWorkingYears",
    "YearsAtCompany",
    "YearsInCurrentRole"
]

X_cluster = df_selected[clustering_features].copy()

print("Clustering data shape:", X_cluster.shape)
print("\nFirst five rows:")
display(X_cluster.head())

print("\nMissing values:")
print(X_cluster.isnull().sum())
```

```python
from sklearn.preprocessing import StandardScaler

scaler_cluster = StandardScaler()

X_cluster_scaled = scaler_cluster.fit_transform(X_cluster)

print("Scaled data shape:", X_cluster_scaled.shape)

print("\nScaled feature means:")
print(pd.Series(
    X_cluster_scaled.mean(axis=0),
    index=clustering_features
).round(3))

print("\nScaled feature standard deviations:")
print(pd.Series(
    X_cluster_scaled.std(axis=0),
    index=clustering_features
).round(3))
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

k_values = range(2, 11)

inertias = []
silhouette_scores = []

for k in k_values:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    cluster_labels = kmeans.fit_predict(X_cluster_scaled)

    inertias.append(kmeans.inertia_)
    silhouette_scores.append(
        silhouette_score(X_cluster_scaled, cluster_labels)
    )

# Elbow method
plt.figure(figsize=(7, 4))
plt.plot(list(k_values), inertias, marker="o")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")
plt.xticks(list(k_values))
plt.tight_layout()
plt.show()

# Silhouette scores
plt.figure(figsize=(7, 4))
plt.plot(list(k_values), silhouette_scores, marker="o")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Silhouette Score")
plt.title("Silhouette Score by Number of Clusters")
plt.xticks(list(k_values))
plt.tight_layout()
plt.show()

# Display scores
cluster_evaluation = pd.DataFrame({
    "K": list(k_values),
    "Inertia": inertias,
    "Silhouette_Score": silhouette_scores
})

display(cluster_evaluation.round(4))
```

```python
from sklearn.cluster import KMeans

final_kmeans = KMeans(
    n_clusters=2,
    random_state=42,
    n_init=10
)

cluster_labels = final_kmeans.fit_predict(X_cluster_scaled)

# Add cluster labels to a copy of the selected dataset
df_clustered = df_selected.copy()
df_clustered["Cluster"] = cluster_labels

print("K-Means model trained successfully!")

print("\nCluster sizes:")
print(df_clustered["Cluster"].value_counts().sort_index())

print("\nSample clustered employees:")
display(
    df_clustered[
        clustering_features + ["Cluster"]
    ].head(10)
)
```

```python
cluster_profile = (
    df_clustered
    .groupby("Cluster")[clustering_features]
    .mean()
    .round(2)
)

print("Average employee characteristics by cluster:")
display(cluster_profile)
```

```python
print("Attrition counts by cluster:")
display(pd.crosstab(
    df_clustered["Cluster"],
    df_clustered["Attrition"]
))

print("\nAttrition percentage within each cluster:")
display(
    pd.crosstab(
        df_clustered["Cluster"],
        df_clustered["Attrition"],
        normalize="index"
    ).mul(100).round(2)
)
```

```python
cluster_profile = (
    df_clustered
    .groupby("Cluster")[clustering_features]
    .mean()
    .round(2)
)

display(cluster_profile)
```

```python

import json
from pathlib import Path

import joblib
import pandas as pd


# Project paths
PROJECT_ROOT = Path(__file__).resolve().parent.parent
MODELS_DIR = PROJECT_ROOT / "models"


def load_models():
    """Load saved employee ML models and configuration."""
    regression_model = joblib.load(
        MODELS_DIR / "monthly_income_regression.joblib"
    )

    classification_model = joblib.load(
        MODELS_DIR / "employee_attrition_classifier.joblib"
    )

    with open(
        MODELS_DIR / "classification_config.json",
        "r",
        encoding="utf-8",
    ) as file:
        config = json.load(file)

    return regression_model, classification_model, config


def predict_employee(employee_data):
    """
    Predict monthly income and attrition for one employee.

    employee_data must contain the same input columns used during
    training, excluding MonthlyIncome and Attrition.
    """
    regression_model, classification_model, config = load_models()

    employee_df = pd.DataFrame([employee_data])

    # Income model requires the same feature columns as X_reg
    income_prediction = regression_model.predict(employee_df)[0]

    # Attrition model uses all 13 features, including MonthlyIncome
    # Use predicted income as the income input for a new employee.
    employee_df["MonthlyIncome"] = income_prediction

    attrition_probability = classification_model.predict_proba(
        employee_df
    )[0]

    yes_index = list(
        classification_model.named_steps["model"].classes_
    ).index("Yes")

    probability_yes = attrition_probability[yes_index]
    threshold = config["decision_threshold"]

    attrition_prediction = (
        "Yes" if probability_yes >= threshold else "No"
    )

    return {
        "predicted_monthly_income": round(
            float(income_prediction), 2
        ),
        "attrition_probability": round(
            float(probability_yes), 4
        ),
        "predicted_attrition": attrition_prediction,
        "decision_threshold": threshold,
    }


if __name__ == "__main__":
    # Example employee input: replace values to test other cases.
    example_employee = {
        "Age": 30,
        "DistanceFromHome": 5,
        "Education": 3,
        "JobLevel": 1,
        "NumCompaniesWorked": 2,
        "TotalWorkingYears": 6,
        "YearsAtCompany": 3,
        "YearsInCurrentRole": 2,
        "BusinessTravel": "Travel_Rarely",
        "Department": "Research & Development",
        "JobRole": "Research Scientist",
        "OverTime": "No",
    }

    result = predict_employee(example_employee)

    for key, value in result.items():
        print(f"{key}: {value}")
```
